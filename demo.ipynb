# Agentic AI Stow Space Optimization Demo
This notebook demonstrates the core optimization and LangGraph AI agent workflow for the Capstone Project.

```python
import sys
import os
from dotenv import load_dotenv

# Ensure the backend module is in the Python path
sys.path.append(os.path.abspath('./backend'))

# Load environment variables (Groq API Key)
load_dotenv('./backend/.env')
```

## 1. Define the Incoming Item
We create a mock item that represents a package arriving at the inbound station.

```python
from app.models.item import ItemInput

item = ItemInput(
    item_id='DEMO-01', 
    length_cm=20.0, 
    width_cm=15.0, 
    height_cm=10.0, 
    weight_kg=2.5, 
    category='general'
)
print(f'Item Dimensions: {item.length_cm}x{item.width_cm}x{item.height_cm} cm | Weight: {item.weight_kg} kg')
```

## 2. Run the Agentic Workflow
We pass the item to the LangGraph agent (`agent_app`). The graph will:
1. Validate the item.
2. Fetch available stow locations from the database.
3. Run the geometric `best_fit` optimization engine.
4. Use the Groq LLM to generate a human-readable explanation.

```python
from app.agent.graph import agent_app
from app.agent.state import StowState

initial_state = StowState(
    item_data=item.model_dump(),
    available_locations=[],
    optimization_result={},
    llm_explanation=''
)

print('Invoking LangGraph Agent workflow...\n')
final_state = agent_app.invoke(initial_state)

print('--- OPTIMIZATION METRICS ---')
print(f"Recommended Location: {final_state['optimization_result'].get('recommended_location')}")
print(f"Space Utilisation: {final_state['optimization_result'].get('space_utilisation_percent')}%")
print(f"Confidence: {final_state['optimization_result'].get('confidence')}")

print('\n--- GROQ LLM EXPLANATION ---')
print(final_state['llm_explanation'])
```
